Draw a systems diagram for the relationship between global mean

PROBLEM 1 ANSWER: The atmospheric CO2 has a positive effect on global mean temperature because higher concentrations strengthen the greenhouse effects. Temperature has a neagtive effect on ocean CO2 concentration through Henry's law, since warmer water holds less dissolved CO2, reducing uptake or increasing outgassing. Amospheric CO2 has a positive effects on ocean CO2 because higher air concentrations create greater disssolution into the ocean. Ocean CO2 also has a positive effect on atmospheric CO2, since higher concentrations can lead to outgassing that increases atmospheric levels.

Together, these relationships form a reinforcing cycle. Rising atmostpheric CO2 raises temperature, which decreases the ocean's ability to absorb CO2, causing more CO2 to remain in or return to the atmosphere, which further raises temperature. This shows a positive feedback between temperature and the ocean carbon cycle where warming diminishes the ocean's role as a carbon sink and instead promotes additional atomspheric CO2 accumulation, increasing climate change.

influence on melting the hypothesized [“Snowball

This simple model of the energy balance (averaged over the entire
planet) is:

<span id="eq-climate">$$
\underbrace{C\frac{dT}{dt}}_{\text{change in heat}} = \underbrace{\frac{(1-\alpha)S}{4}}_{\text{incoming radiation}} - \underbrace{(A - BT)}_{\text{outgoing radiation}} + \underbrace{a\ln \left(\frac{[CO_2]}{[CO_2]_{PI}}\right)}_{\text{greenhouse effect}},
 \qquad(1)$$</span>

where:

-   $T$ is the Earth’s global mean temperature (in $^\circ\text{C}$);
-   $C$ is the heat capacity of the atmosphere and shallow ocean, taken
    to be $51\ \text{J}/\text{m}^2/^\circ\text{C}$;
-   $\alpha$ is the planetary albedo, or the fraction of incoming
    radiation reflected by the Earth, which has a present-day value of
    approximately 0.3;
-   $S$ is the solar constant, or the amount of solar radiation received
    by the Earth averaged over area, which has a present-day value of
    $1368\ \text{W}/\text{m}^2$ but during the Neoproteorozoic Era had a
    value of $1272\ \text{W}/\text{m}^2$ (this is divided by four

-   $A$ and $B$ are coefficients from the linearization of outgoing

    $A=221.2\ \text{W}/\text{m}^2$ (estimated by assuming the
